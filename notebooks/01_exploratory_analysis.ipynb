# Exploratory Analysis of Antarctic Soundscapes

This notebook performs an exploratory analysis on field recordings of the Antarctic soundscape using the `soundsanity` library. We will load the audio examples, plot their waveforms, and run basic sanity check metrics to establish baseline values for clean recordings.

### 1. Imports and setup

```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
import essentia.standard as es
import soundsanity as ss

# Set inline plotting style
%matplotlib inline
plt.rcParams["figure.figsize"] = (12, 4)
plt.style.use("seaborn-v0_8-whitegrid")
```

### 2. Locate and Inspect Audio Examples

We will search the `audio_examples` folder for `.wav` and `.mp3` files.

```python
audio_dir = "../audio_examples"
wav_files = sorted(glob.glob(os.path.join(audio_dir, "*.wav")))
mp3_files = sorted(glob.glob(os.path.join(audio_dir, "*.mp3")))
all_files = wav_files + mp3_files

print(f"Found {len(wav_files)} WAV files and {len(mp3_files)} MP3 files.")
for f in all_files:
    print(f"  - {os.path.basename(f)}")
```

### 3. Run Batch Quality Diagnostics

We will analyze all files and display their diagnostic reports.

```python
reports = []
for f_path in all_files:
    report = ss.analyze_recording(f_path)
    reports.append(report)
    
    print(f"File: {report['file_name']}")
    print(f"  Duration: {report['duration']:.2f} seconds")
    print(f"  Status: {report['status']}")
    print(f"  Clicks count: {report['clicks']['clicks_count']}")
    print(f"  Saturation %: {report['saturation']['saturation_ratio'] * 100:.4f}%")
    print(f"  Silence ratio: {report['silence']['silence_ratio']:.2%}")
    print(f"  Noise Floor: {report['noise']['noise_floor_db']:.1f} dB")
    print(f"  SNR index: {report['noise']['snr_db']:.1f} dB")
    print("-" * 50)
```

### 4. Detailed Visual Reports for Key Recordings

Let's visualize the audio properties of two distinct recordings: `pinguinos1.wav` (which contains vocalizations) and `deshielo.wav` (which contains melting ice sounds).

```python
# 1. Penguins Recording
ping_file = os.path.join(audio_dir, "pinguinos1.wav")
ping_audio = es.MonoLoader(filename=ping_file, sampleRate=44100)()
ping_report = ss.analyze_recording(ping_file)

fig, axs = ss.plot_quality_report(ping_audio, 44100, ping_report, title="Penguin Soundscape")
```

```python
# 2. Ice Melting Recording
ice_file = os.path.join(audio_dir, "deshielo.wav")
ice_audio = es.MonoLoader(filename=ice_file, sampleRate=44100)()
ice_report = ss.analyze_recording(ice_file)

fig, axs = ss.plot_quality_report(ice_audio, 44100, ice_report, title="Melting Ice")
```

### 5. Conclusions

All the provided recordings are of high quality, with no clipping/saturation, and very low click counts. The noise levels are typical of outdoor soundscapes.
